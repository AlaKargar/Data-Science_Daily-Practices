```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = ['فروردین', 'اردیبهشت', 'خرداد', 'تیر', 'مرداد', 'شهریور']
sells = [150, 200, 250, 300, 400, 350]
sells2 = [200, 150, 300, 450, 350, 250]
```

```python
plt.scatter(months, sells, color='red', s=100, alpha=0.7)
plt.plot(months, sells, ".", ls = ":", color = "purple")

plt.scatter(months, sells2, color='blue', s=100, alpha=0.7)
plt.plot(months, sells2, "*", ls = "-", color = "green")

plt.xlabel("Months")
plt.ylabel("sells")
plt.title("Sells in months")

plt.grid()
```

```python
x = np.random.rand(50) * 8
y = np.random.rand(50) * 8

fig, ax = plt.subplots()
ax.scatter(x, y)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```

```python
plt.bar(months, sells, color = "pink", width= 0.3 )

plt.xlabel("Months")
plt.ylabel("sells")
plt.title("Sells in months")
```

```python
mycolors = ["purple", "blue", "green", "yellow", "orange", "red"]
plt.pie(sells, colors=mycolors, startangle=90)
plt.legend(months)
plt.title("Sells in months")
```
