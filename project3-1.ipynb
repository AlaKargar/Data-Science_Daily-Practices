```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2, 5, 7])
y = np.array([3,6, 9])
plt.scatter(x,y)
plt.plot(x,y)

z = ([2, 4, 6])
w = ([4, 7, 10])
plt.scatter(z, w)
plt.plot(z, w)

plt.xlabel("Hours of studying")
plt.ylabel("Days of studying")
plt.title("STUDYING:")

plt.show()
```
