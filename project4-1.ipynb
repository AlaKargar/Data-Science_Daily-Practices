```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6],
    "Test_Score": [50, 60, 65, 70, 75, 85]
}

df = pd.DataFrame(data)

sns.scatterplot(x="Hours_Studied", y="Test_Score", data=df)
plt.title("Relationship Between Study Hours and Test Scores")
plt.show()
```

```python
hours_studies = data['Hours_Studied']
test_score = data['Test_Score']

plt.scatter(hours_studies, test_score, color='pink', label = 'Test score')

plt.title("Relationship Between Study Hours and Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Test Score")
plt.legend()
```
