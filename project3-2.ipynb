```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([1, 2, 3])
y = np.array([1, 6, 5])
```

```python
z = np.array([1, 2, 3])
w = np.array([2, 4, 4])
```

```python
plt.plot(x, y, ".", ls = ":", color = "purple")
plt.plot(z, w, ".", ls = ":", color = "red")

plt.xlabel("DAYS OF DRINKING")
plt.ylabel("GLASSES OF WATER")
plt.title("DRINKING WATER: ")
plt.grid()
```
