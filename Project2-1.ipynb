```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
numbers=np.array([90,10])
when=["After sending the massage","Before sending the message"]
```

```python
mycolors=["green","blue"]
plt.pie(numbers, colors=mycolors, startangle=90)
plt.legend(labels=when)
plt.title("When you figured out that you have written the message in wrong way!")
```
