```python
import pandas as pd
import numpy as np
```

```python
weights = np.random.randint(10, 50, 10)
df = pd.DataFrame({'weight':weights})

mean_weights = df['weight'].mean()
max_weights = df['weight'].max()
min_weights = df['weight'].min()

print("mean weights", mean_weights)
print("max weights", max_weights)
print("min weights", min_weights)
```
